```python
# Always import packages first
import pandas as pd
import numpy as np
```

```python
# A pandas.series having an index is the main difference between a pandas.series and a NumPy array.

# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# To create a pandas.series you use pd.series

# s = pd.series(data, index = index)

# 'data' parameter can be a list or NumPy array, a python dictionary or a single number, boolean, or string

# the index parameter is optional. if you want to include it, it has to be a list of indices the same length as data

```

```python
# Creating pandas.series from numpy array

pd.Series(np.arange(3), index=[2023,2024,2025])
```

```python
# Creating a pandas.series from a list

pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# You can even create one from a dictionary

d = {'key_0':2, 'key_1':'3', 'key_2':5}
pd.Series(d)
```

```python
# Creating pandas.series from a single value DEEPLY USEFUL

pd.Series(3.0, index = ['A','B','C'])
```

```python
# Even arithmetic operations work on series

s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

s/10

np.exp(s)
```

```python
# Now, what about booleans? They are evaluated for each row

s > 70
```

```python
# Finding na's in a pd.series

s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# How to check for NA's

s.hasnans

# .hasnans says if there are ANY NA's in a series/df
```

```python
# now that we know there are NA's, we can check which rows have them

s.isna()
```

```python
# Check-in 1

nines = pd.Series([1,2,-999,-999], index= ['A','B','C','D'])

nines.mask(nines < 0)
```

```python
# Creating a pandas df using dictionaries

d = {'col_name_1' : pd.Series(np.arange(3)),
    'col_name_2' : pd.Series([3.1,3.2,3.3])}

df = pd.DataFrame(d)
df
```

```python
# Can change index using df.Index

df.index = ['a','b','c']
df
```
