```python
import pandas as pd
import numpy as np
import os
```

```python
path = os.path.join('data','wetlands_seasonal_bird_diversity.csv')
bird = pd.read_csv(path)
```

```python
bird.head()
```

```python
bird.plot()
```

```python
# Formula for adding values
# bird.plot(x = 'x_values_column', y = 'y_values_calumn')

bird.plot(x='year', y = 'CSM_winter')
```

```python
bird.plot(x='year', 
        y='CSM_winter',
        title='Bird species registered during winter at Carpinteria Salt Marsh',
        xlabel='Year',
        ylabel='Number of bird species',
        color='green',
        legend=False
        )
```

```python
# Check in 1

bird.plot(x = 'year', y = 'MUL_spring')
bird[bird['MUL_spring'].isna()]
```

```python
 # Check in # 2

 # well first it sets the index using the year column,
 # then it locates rows between SDW winter and TJE fall,
 # and plots those guys

 # the code DOES modify df because it resets the index to the year column,
 # but it does not commit those changes so the actual df is unaltered

bird.set_index('year').loc[:,'SDW_winter':'TJE_fall'].plot()

# Well, there are some missing data points for SDW_spring?
```
